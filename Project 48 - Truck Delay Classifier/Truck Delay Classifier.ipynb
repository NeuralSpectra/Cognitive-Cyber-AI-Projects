# **Data Cleaning**

**1: The function categorize_time categorizes hours into segments: Early Morning, Morning, Noon, Evening, and Night based on specified ranges.**

**2: A copy of the traffic_table DataFrame is created, named traffic.**

**3: The categorize_time function is applied to the 'hour' column to create a new column, 'time_category', in the traffic DataFrame.**

**4: The mean number of vehicles is calculated for each time category by grouping the data in the traffic DataFrame.**

```python
def categorize_time(hour):

    if 300 <= hour < 600:
        return 'Early Morning'
    elif 600 <= hour < 1200:
        return 'Morning'
    elif 1200 <= hour < 1600:
        return 'Noon'
    elif 1600 <= hour < 2000:
        return 'Evening'
    elif 2000 <= hour < 2300:
        return 'Night'
    elif 0 <= hour < 300:
        return 'Night'

traffic = traffic_table.copy()

traffic['time_category'] = traffic['hour'].apply(categorize_time)

mean_vehicles_by_time = traffic.groupby('time_category')['no_of_vehicles'].mean()
```

**The variable mean_vehicles_by_time contains the average number of vehicles categorized by the time segments defined in the categorize_time function.**

```python
mean_vehicles_by_time
```

**Fill missing values in driving_style: This line replaces any NaN values in the driving_style column of the drivers_table DataFrame with the string 'Unknown', ensuring that all entries have a value.**

**Fill missing values in gender: Similar to the previous line, this replaces NaN values in the gender column with 'Unknown', maintaining data integrity.**

**Replace empty strings in fuel_type: This line replaces any empty strings in the fuel_type column of the trucks_table_df DataFrame with 'Unknown', ensuring that the column does not contain any empty values.**

```python
drivers_table['driving_style'] = drivers_table['driving_style'].fillna('Unknown')
drivers_table['gender'] = drivers_table['gender'].fillna('Unknown')
trucks_table_df['fuel_type'] = trucks_table_df['fuel_type'].replace("",'Unknown')
```

# **Model Building**

**In this code, you're defining the feature set X and the target variable y for a machine learning model:**

```python
X = df[['precip_x', 'humidity_x', 'visibility_x', 'pressure_x', 'precip_y',
       'humidity_y', 'visibility_y', 'pressure_y', 'distance', 'average_hours',
       'no_of_vehicles', 'Day', 'Month', 'Year']]

y = df[["accident"]]
```

**The code applies Min-Max Scaling to the feature set (X) and target variable (y) using MinMaxScaler from sklearn.preprocessing. It initializes scalers for both X and y, then fits and transforms them to ensure all values are scaled to a range of 0 to 1. This normalization helps improve the performance of machine learning models sensitive to data scales.**

```python
from sklearn.preprocessing import MinMaxScaler

Scaler_X = MinMaxScaler().set_output(transform = "pandas")

X = Scaler_X.fit_transform(X)

Scaler_y = MinMaxScaler().set_output(transform = "pandas")

y = Scaler_y.fit_transform(y)
```

**This code imports train_test_split and classification_report from sklearn. The train_test_split function is used to divide the dataset into training and testing subsets, while classification_report generates a report that shows the main classification metrics, such as precision, recall, and F1-score, for evaluating model performance.**

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

**This line of code splits the dataset into training and testing sets, allocating 80% of the data for training (X_train and y_train) and 20% for testing (X_test and y_test). This separation helps in evaluating the model's performance on unseen data.**

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.80)
```

**1: Model Initialization: A Logistic Regression model is created with fit_intercept set to False and parallel processing enabled with n_jobs=-1.**

**2: Model Training: The model is trained using the training data (X_train and y_train), allowing it to learn the relationship between the features and the target variable.**

**3: Prediction: Predictions are made on the test dataset (X_test) to assess the model's performance on unseen data.**

**4: Classification Report: The classification report, which includes precision, recall, and F1-score for each class, is generated and printed to evaluate the model's effectiveness.**

```python
from sklearn.linear_model import LogisticRegression

LogReg = LogisticRegression(fit_intercept=False, n_jobs=-1)

LogReg.fit(X_train, y_train)

y_pred = LogReg.predict(X_test)

CR = classification_report(y_pred, y_test)

print("Classification Report For Logistic Regression: ")
print("")
print(CR)
```

**1: Model Initialization: A Random Forest Classifier is initialized with 100 trees (n_estimators=100), a maximum depth of 10 for each tree (max_depth=10), and class balancing enabled to address any class imbalance in the dataset.**

**2: Model Training: The model is trained on the training dataset (X_train and y_train), where it learns to make predictions based on the features provided.**

**3: Prediction: Predictions are generated for the test dataset (X_test) to evaluate how well the model performs on unseen data.**

**4: Classification Report: A classification report is created to summarize the model's performance, displaying metrics such as precision, recall, and F1-score for each class, and it is then printed to the console.**

```python
from sklearn.ensemble import RandomForestClassifier

RFC = RandomForestClassifier(n_estimators=100, max_depth=10, n_jobs=-1, class_weight='balanced')

RFC.fit(X_train, y_train)

y_pred = RFC.predict(X_test)

CR = classification_report(y_pred, y_test)

print("Classification Report For Random Forest Classifier: ")
print("")
print(CR)
```

**1: Model Initialization: A Decision Tree Classifier is initialized with the entropy criterion for measuring the quality of a split, a maximum depth of 5 (max_depth=5), a minimum sample split of 50 (min_samples_split=50), and a minimum leaf size of 25 samples (min_samples_leaf=25).**

**2: Model Training: The model is trained on the training dataset (X_train and y_train), where it learns to partition the data based on the features provided to make predictions.**

**3: Prediction: Predictions are made for the test dataset (X_test), allowing the model to assess its ability to classify unseen data.**

**4: Classification Report: A classification report is generated to evaluate the model's performance, detailing metrics such as precision, recall, and F1-score for each class, which is then printed to the console.**

```python
from sklearn.tree import DecisionTreeClassifier

DTC = DecisionTreeClassifier(criterion='entropy', max_depth=5, min_samples_split=50, min_samples_leaf=25)

DTC.fit(X_train, y_train)

y_pred = DTC.predict(X_test)

CR = classification_report(y_pred, y_test)

print("Classification Report For Decision Tree Classifier: ")
print("")
print(CR)
```

**The Sequential model and Dense layers are imported from TensorFlow's Keras to build neural networks. The Sequential model allows stacking layers linearly, enabling the creation of complex architectures.**

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

**This code defines a Sequential neural network model with four layers:**

- **Input Layer: The first layer has 512 neurons, uses ReLU activation, and expects 14 input features.**

- **Hidden Layers: The second and third layers have 256 and 128 neurons respectively, both using ReLU activation.**

- **Output Layer: The final layer has 1 neuron and also uses ReLU activation, indicating it outputs a single continuous value.**

- **Summary: model.summary() displays a summary of the model's architecture, including the number of parameters and layer configurations.**

```python
model = Sequential()

model.add(Dense(512, activation = "relu", input_dim = 14))
model.add(Dense(256, activation = "relu"))
model.add(Dense(128, activation = "relu"))
model.add(Dense(1, activation = "sigmoid"))

model.summary()
```

**This code snippet performs the following actions on the X_train DataFrame:**

- **Replace Infinite Values: It replaces any occurrences of positive infinity (np.inf) and negative infinity (-np.inf) in X_train with NaN (Not a Number) using replace().**

- **Drop Missing Values: It removes any rows that contain NaN values using dropna(), ensuring that the training data does not have any missing or infinite values for the model training process.**

```python
X_train.replace([np.inf, -np.inf], np.nan, inplace=True)
X_train.dropna(inplace=True)
```

**This line of code compiles the neural network model with the following specifications:**

- **Loss Function: It uses "categorical_crossentropy" as the loss function, which is suitable for multi-class classification problems where the target variable is one-hot encoded.**

- **Optimizer: The model employs the "adam" optimizer, a popular choice due to its adaptive learning rate capabilities, which can improve training efficiency.**

- **Metrics: It specifies "accuracy" as a metric to evaluate the model's performance during training and validation, allowing for monitoring of the model's accuracy on the dataset.**

```python
model.compile(loss = "categorical_crossentropy", optimizer = "adam", metrics=["accuracy"])
```

**This line of code trains the neural network model with the following parameters:**

- **Training Data: It uses X_train as the input features and y_train as the target labels.**

- **Epochs: The model is trained for 5 epochs, meaning the training dataset will be passed through the model 5 times.**

- **Batch Size: The batch size is set to 7000, indicating that the model will process 7000 samples at a time before updating the weights.**

- **Validation Split: A validation split of 0.20 (20%) is used, which means 20% of the training data will be set aside for validation during training to monitor the model's performance.**

```python
model.fit(X_train, y_train, epochs=5, batch_size = 7000, validation_split=0.20)
```
