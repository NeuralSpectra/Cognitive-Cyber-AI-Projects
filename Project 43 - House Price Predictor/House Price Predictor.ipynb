# **Data Visualization**

**The code defines a function that analyzes and visualizes the impact of missing values in each column on the logerror variable by plotting the median logerror values for missing and non-missing entries.**

**It then applies this function to all columns with missing values in the DataFrame.**

**In this graph, we are analyzing all the null values in relation to the 'logerror' column, which serves as the target variable.**

```python
def analyze_missing_values(df, var):

    print("")
    print("*" * 75)
    print("")
    df[var] = np.where(df[var].isnull(), 1, 0)
    df.groupby(var)["logerror"].median().plot.bar()
    plt.title(var)
    plt.show()

for var in df.columns[df.isnull().sum() > 0]:
    analyze_missing_values(df, var)
```

**analyse_log_variation(df, var): This function performs a logarithmic transformation on the specified continuous variable var if all its values are positive. It then creates a histogram of the transformed values. If the variable contains non-positive values, it skips the transformation and prints a message.**

**Loop: Iterates through each continuous variable in continues_values and applies analyse_log_variation to visualize the effect of the logarithmic transformation.**

**In this code, we analyze the logarithmic transformation and variation for each column, except for columns with values less than 0.**

```python
def analyse_log_variation(df, var):

    if any(df[var] <= 0):
        print(f"Skipping Logarithmic Transformation For {var} Due To Non-Positive Values.")
        return
    else:
        df[var] = np.log(df[var])
        df[var].hist(bins=20)
        print("\n" + "*" * 75 + "\n")
        plt.xlabel(var)
        plt.ylabel("Count")
        plt.title("Logarithmic Variation Of: " + var)
        plt.show()


for var in continues_values:
    analyse_log_variation(df, var)
```

**analyse_outliers(df, var): This function checks if the specified continuous variable var contains non-positive values. If it does, it skips the logarithmic transformation and prints a message. Otherwise, it applies the logarithmic transformation and creates a boxplot to visualize the distribution and potential outliers in the transformed data.**

**Loop: Iterates through each continuous variable in continues_values and applies analyse_outliers to visualize the effect of the logarithmic transformation on outliers.**

**In this code, we analyze the outliers present in each column and remove them, as they can negatively impact the accuracy of machine learning algorithms.**

```python
def analyse_outliers(df, var):

    if any(df[var] <= 0):
        print(f"Skipping Logarithmic Transformation For {var} Due To Non-Positive Values.")
    else:
        print("\n" + "*" * 75 + "\n")
        df[var] = np.log(df[var])
        df.boxplot(column=var)
        plt.title(f"Boxplot For Logarithmic Variation Of {var}")
        plt.show()


for var in continues_values:
    analyse_outliers(df, var)
```

**Print Missing Values: Displays the count of missing values for columns listed in missing_values to assess their completeness.**

**Set Threshold: Defines a limit as 60% of the total number of rows to identify columns with a high percentage of missing values.**

**Identify Columns to Drop: Creates a list of column names where the count of missing values exceeds the defined threshold and prints these columns.**

**Drop and Update DataFrame: Removes columns with more than 60% missing values from the DataFrame, prints the remaining columns, and returns the updated DataFrame.**

```python
def drop_columns_with_max_missing_values(df):

    print(df[missing_values].isnull().sum())

    limit = np.abs((df.shape[0] * 0.6))

    values_to_drop = [var for var in missing_values if df[var].isnull().sum() > limit]

    print("Columns With More Than 60% Missing Values: \n\n", values_to_drop)
    print("")
    print("Dropping Columns With More Than 60% Missing Values: \n")

    df.drop(columns = values_to_drop, axis=1, inplace = True)

    print("Remaining Columns: \n", df.columns)

    return df

df = drop_columns_with_max_missing_values(df)
```

```python
X = df[['airconditioningtypeid', 'architecturalstyletypeid',
       'basementsqft', 'bathroomcnt', 'bedroomcnt', 'buildingclasstypeid',
       'buildingqualitytypeid', 'decktypeid', 'finishedfloor1squarefeet',
       'finishedsquarefeet12', 'finishedsquarefeet13', 'finishedsquarefeet15',
       'finishedsquarefeet50', 'finishedsquarefeet6', 'fips', 'fireplacecnt',
       'garagecarcnt', 'garagetotalsqft', 'hashottuborspa',
       'heatingorsystemtypeid', 'latitude', 'longitude', 'lotsizesquarefeet',
       'poolcnt', 'poolsizesum', 'pooltypeid10', 'pooltypeid2', 'pooltypeid7',
       'propertycountylandusecode', 'propertylandusetypeid',
       'propertyzoningdesc', 'rawcensustractandblock', 'regionidcity',
       'regionidcounty', 'regionidneighborhood', 'regionidzip', 'roomcnt',
       'storytypeid', 'threequarterbathnbr', 'typeconstructiontypeid',
       'unitcnt', 'yardbuildingsqft17', 'yardbuildingsqft26',
       'numberofstories', 'fireplaceflag', 'taxamount', 'taxdelinquencyflag',
       'taxdelinquencyyear', 'yeardifference']]

y = df[['logerror']]
```

**Initialize Scalers: Create MinMaxScaler instances for both features (X) and target (y), setting transform="pandas" to ensure output as pandas DataFrames.**

**Fit and Transform Features: Scale the feature set X to the range [0, 1] using Scaler_X.**

**Fit and Transform Target: Scale the target variable y to the range [0, 1] using Scaler_y.**

```python
from sklearn.preprocessing import MinMaxScaler

Scaler_X = MinMaxScaler().set_output(transform = "pandas")

X = Scaler_X.fit_transform(X)

Scaler_y = MinMaxScaler().set_output(transform = "pandas")

y = Scaler_y.fit_transform(y)
```

# **Deep Learning**

**Import Layers: Import the Dense layer from tensorflow.keras.layers to define fully connected layers in a neural network.**

**Import Model: Import the Sequential model from tensorflow.keras.models to create a linear stack of layers for building neural network models.**

```python
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential
```

**Initialize Model: Create a Sequential model, which allows for stacking layers in a linear fashion.**

**Add Layers:**

- **Input Layer: Dense(1024, activation="relu", input_dim=49) initializes a fully connected layer with 1024 units and ReLU activation, accepting an input with 49 features.**

- **Hidden Layers: Dense(512, activation="relu") and Dense(256, activation="relu") add fully connected layers with 512 and 256 units respectively, both using ReLU activation.**

- **Output Layer: Dense(1, activation='linear') adds the final layer with 1 unit and a linear activation function, suitable for regression tasks.**

```python
model = Sequential()

model.add(Dense(1024, activation="relu", input_dim = 49))
model.add(Dense(512, activation="relu"))
model.add(Dense(256, activation="relu"))
model.add(Dense(1, activation='linear'))
```

**The model.summary() method provides a detailed overview of the neural network model, including:**

**Layer Information: Displays each layer's type, output shape, and number of parameters.**

**Parameter Counts: Shows the total number of trainable and non-trainable parameters in the model.**

**Model Architecture: Summarizes the structure of the model, including the input shape and connections between layers.**

```python
model.summary()
```

**Optimizer: 'adam' is used to optimize the model, which adapts the learning rate and helps in efficient training.**

**Loss Function: 'mean_squared_error' is specified for regression tasks to measure the average squared difference between predicted and actual values.**

**Metrics: ['accuracy'] is included to track the model's performance during training and evaluation, although accuracy is more commonly used for classification tasks. For regression, you might consider using metrics like mean_absolute_error or mean_squared_error for more relevant performance tracking.**

```python
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['accuracy'])
```

**Train the Model: model.fit(X_train, y_train, epochs=10, batch_size=128, validation_split=0.20) trains the neural network on the training data.**

**Epochs: epochs=10 specifies the number of complete passes through the training data.**

**Batch Size: batch_size=128 sets the number of samples per gradient update.**

**Validation Split: validation_split=0.20 reserves 20% of the training data for validation to monitor the model's performance on unseen data during training.**

```python
history = model.fit(X_train, y_train, epochs=10, batch_size=128, validation_split=0.20)
```
